# Getting Halo Initial Conditions from Stern+19

```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.constants as const
import scipy.integrate as integrate
import math
```

```python
plt.rcParams['font.size']=22
plt.rcParams['axes.linewidth']=2
plt.rcParams['xtick.major.size']=10
plt.rcParams['xtick.minor.size']=5
plt.rcParams['xtick.major.width']=2
plt.rcParams['xtick.minor.width']=1
plt.rcParams['xtick.direction']='in'
plt.rcParams['ytick.major.size']=10
plt.rcParams['ytick.minor.size']=5
plt.rcParams['ytick.major.width']=2
plt.rcParams['ytick.minor.width']=1
plt.rcParams['ytick.direction']='in'
```

```python
#Equation 1
#dM_dt = 4 * pi * r*r * rho * v

#Equation 2
#vdv_dr = - (1/rho) * dP_dr - vc*vc/r

#Equation 3
#vdlnK_dr = -1/tcool
```

```python
#nH = X * rho/mp
#tcool = P/(gamma-1)/nH/nH/Lambda
#vc = np.sqrt(G*Mgrav/r)
#Mgrav = NFW + Stellar + Outer
```

```python
#Constants
H0   = 67.0
Msun = const.M_sun.cgs.value
G = const.G.cgs.value
mp = const.m_p.cgs.value
kpc = const.kpc.cgs.value
```

```python
def c(Mhalo):
    h_inv = H0/100.0
    log10_c = 1.025 - 0.097 * np.log10(Mhalo/1.e12/Msun/h_inv)
    return 10.**log10_c

def Rvir(Mhalo):
    R_vir = 260.0  * (Mhalo/1.e12/Msun)**(1./3.) * kpc
    return R_vir

def NFW(r, Rvir, Mhalo):
    Rs = Rvir/c(Mhalo)
    rho_r = 1/(r/Rs)/(1+r/Rs)**2
    return rho_r * 4. * math.pi * r * r

def NFW_rho(r, Rvir, Mhalo):
    Rs = Rvir/c(Mhalo)
    rho_r = 1/(r/Rs)/(1+r/Rs)**2
    return rho_r 

def mass_enc_halo(r, Rvir, Mhalo, rho0):
    Rs = Rvir/c(Mhalo) 
    rho_r = rho0/(r/Rs)/(1+r/Rs)**2
    dm = 4. * math.pi * r * r * rho_r
    return dm
```

```python
Mhalo = 1.e10 * Msun
Mstar = 0.01 * Mhalo
Rvir_ = Rvir(Mhalo)
rad = np.linspace(0.01, 2.0*Rvir_/kpc, 1000)
I_NFW = integrate.quad(NFW, 0, Rvir_, args=(Rvir_, Mhalo))
central_rho = Mhalo/I_NFW[0]
```

```python
np.amax(rad), Rvir_
```

```python
rho_halo = NFW_rho(rad*kpc, Rvir_, Mhalo )

plt.plot(rad, rho_halo, label='NFW')
plt.yscale('log')
plt.xscale('log')
plt.ylabel(r'$\rho/\rho_0$')
plt.xlabel('r [kpc]')
plt.ylim(1.e-4,1.e3)
plt.axvline(Rvir_/kpc,ls='--', color='red', label=r'$R_{\rm vir}$')
plt.legend()
plt.savefig('nfw.jpeg', bbox_inches='tight')
```

```python
#Estimate mass enclosed in halo
mass_enclosed_nfw = np.zeros(rad.shape)
i=0
for r in rad:
    mass_enclosed_nfw[i] = integrate.quad(mass_enc_halo, 0.001, r*kpc, args=(Rvir_, Mhalo, central_rho))[0]
    i+=1
```

```python
#Stellar Mass
def R_half(Mhalo):
    r = 3.0 * (Mhalo/1.e12/Msun) **(1./3.) * kpc
    return r

def M_star_enc(r, Mhalo):
    Mstar = 0.01 * Mhalo
    enc_stellar_mass = Mstar * r/(r + R_half(Mhalo))
    return enc_stellar_mass

M_star = M_star_enc(rad*kpc, Mhalo)

plt.plot(rad, M_star)
plt.yscale('log')
plt.xscale('log')
plt.ylabel(r'$M_* (r<R)$')
plt.xlabel('r [kpc]')
```

```python
#Outer component

def mass_enc_outer(r, Mhalo):
    R200 = 200. * (Mhalo/1.e12/Msun) ** (1./3.) * kpc
    rho_m = 2.88e-30 #g/cm3
    be = 1.0
    se = 1.5
    rho = rho_m * (be * (r/(5.0*R200))**(-se) + 1)
    dm = (4. * math.pi * r * r * rho)
    return dm 

mass_enclosed_outer = np.zeros(rad.shape)
i=0
for r in rad:
    mass_enclosed_outer[i] = integrate.quad(mass_enc_outer, 0.001, r*kpc, args=(Mhalo))[0]
    i+=1
```

```python
plt.figure(figsize=(8,8))
total = mass_enclosed_nfw + M_star + mass_enclosed_outer
plt.plot(rad, mass_enclosed_nfw/Msun, label='NFW Halo', lw=4.0)
plt.plot(rad, M_star/Msun, label='Stellar', lw=4.0)
plt.plot(rad, mass_enclosed_outer/Msun, label = 'Outer', lw=4.0)
plt.plot(rad,total/Msun, color='black', label='Total', lw=2.0)
plt.yscale('log')
# plt.xscale('log')
plt.ylabel(r'M ($<r$) [$M_{\odot}$]')
plt.xlabel('r [kpc]')
plt.legend()
plt.ylim(1.e6,3.e10)
plt.axvline(Rvir_/kpc,ls='--', color='red', label=r'$R_{\rm vir}$')
plt.savefig('mass_profiles.jpeg', bbox_inches='tight')
```
