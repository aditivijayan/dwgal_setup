```python
box_low = -600
box_high = 600

# Create a boolean mask: True for particles inside the box
mask = (xhalo > box_low) & (xhalo < box_high) & \
       (yhalo > box_low) & (yhalo < box_high) & \
       (zhalo > box_low) & (zhalo < box_high)

# Apply the mask
x_box = xhalo[mask]
y_box = yhalo[mask]
z_box = zhalo[mask]

vx_box = vx_halo[mask]
vy_box = vy_halo[mask]
vz_box = vz_halo[mask]

mass_box = mass[mask]
print(f"Selected {len(x_box)} particles out of {len(xhalo)}")
```

```python
filename ='LMC_particles_1e7Msun.txt'

Msun = 1.989e+33
kmps = 1.e5
kpc  = 3.018e21

header = str(x_box.shape[0])
data = np.column_stack([x_box*kpc, y_box*kpc, z_box*kpc, mass_box, vx_box*kmps, vy_box*kmps, vz_box*kmps ])
np.savetxt(filename, data, header=header, fmt="%.6e", comments='')
```
